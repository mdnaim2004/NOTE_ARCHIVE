```python
import numpy as np
```

```python
x = np.arange(5,20)
x
```

```python
y = x*5 + 1
y
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(x,y,'r')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.title('the graph of matplotlib implimentation')
plt.show()
```

```python
plt.show() # notheing, cause it endimg point of the bloc of code
# and, plt.show er maddhome nides kore je akta code block er ending point,
```

```python
plt.subplot(1,2,2)
plt.plot(x, y, 'r')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.subplot(1,2,1)
plt.plot(y,x)
plt.xlabel('y axis')
plt.ylabel('x axis')
plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(x,y, 'r--')
plt.subplot(1,2,2)
plt.plot(y,x, 'r+-')
plt.show()
```

```python
# Object Orentade Methode

fig = plt.figure()
fig
```
