```python
import numpy as np
```

```python
x = np.arange(5,20)
x
```

```python
y = x*5 + 1
y
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(x,y,'r')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.title('the graph of matplotlib implimentation')
plt.show()
```
