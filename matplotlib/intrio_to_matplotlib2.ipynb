## ডেটা তৈরি ও সেটআপ

এখানে আমরা প্রথমে `x` ও `y` ডেটা তৈরি করছি।

- `x = np.arange(5, 20)` মানে ৫ থেকে ১৯ পর্যন্ত সংখ্যাগুলো তৈরি হচ্ছে।
- `y = x * 5 + 1` মানে প্রতিটি `x` এর সাথে ৫ গুণ করে ১ যোগ করা হচ্ছে।

একই ডেটা আমরা পরে plotting-এর জন্য ব্যবহার করব।

অর্থাৎ, এখানে আমরা graph করার জন্য প্রস্তুত ডেটা তৈরি করছি।

```python
import matplotlib.pyplot as plt
```

## গ্রাফিং অংশ: কী কী হচ্ছে এবং কী করছি

এই সেকশনে আমরা Matplotlib দিয়ে একাধিক গ্রাফ দেখানোর পদ্ধতি শিখছি।

- `plt.subplot(1, 2, 1)` মানে ১x২ লেআউটে প্রথম subplot তৈরি করা হয়েছে।
- `plt.plot(x, y, 'r--')` দিয়ে x ও y ডেটা নিয়ে লাল dashed লাইন তৈরি করা হয়েছে।
- `plt.subplot(1, 2, 2)` দিয়ে একই লাইনে দ্বিতীয় subplot তৈরি করা হয়েছে।
- `plt.plot(y, x, 'r+-')` এর মাধ্যমে y vs x plot তৈরি করা হয়েছে, যেখানে ডট-ক্রস স্টাইল ব্যবহার হয়েছে।

এখানে আমরা দেখতে পাচ্ছি, একই plot area-এ দুইটি আলাদা graph একসাথে সাজানো হচ্ছে। অর্থাৎ, একই ডেটা বিভিন্নভাবে বিশ্লেষণ করা হচ্ছে।

তারপর `fig = plt.Figure()` দিয়ে একটি কাস্টম figure তৈরি করা হয়।
`fig.add_axes([.1, .1, .8, .8])` দিয়ে figure-এ plot এর আকার ও অবস্থান নির্ধারণ করা হয়।

সারসংক্ষেপ:
- কোডগুলো কোন জায়গা থেকে আসছে: `matplotlib.pyplot` library থেকে।
- কী হচ্ছে: plot, subplot, axes, figure তৈরি করা হচ্ছে।
- কী করছি: গ্রাফের ভিজুয়াল আকার, অবস্থান, রঙ, স্টাইল ও layout নির্ধারণ করছি।

অর্থাৎ, আমি এখানে ডেটা কিভাবে ভিজুয়াল আকারে দেখাচ্ছি সেটি বুঝিয়ে দিচ্ছি।

```python
plt.subplot(1,2,1)
plt.plot(x,y, 'r--')
plt.subplot(1,2,2)
plt.plot(y,x, 'r+-')
plt.show()
```

## subplot দিয়ে দুইটি গ্রাফ একসাথে

এখানে আমরা `plt.subplot(1,2,1)` এবং `plt.subplot(1,2,2)` ব্যবহার করেছি।

- ১x২ মানে একই figure-এ ২টি কলাম থাকবে।
- প্রথম subplot-এ `x` vs `y` লাইন plot হয়েছে।
- দ্বিতীয় subplot-এ `y` vs `x` plot হয়েছে।
- `'r--'` মানে লাল রঙের dashed line।
- `'r+-'` মানে লাল রঙের plus-marked line।

এখানে মূল লক্ষ্য হলো: একই ডেটাকে ভিন্নভাবে সাজিয়ে তুলনা করা।

অর্থাৎ, আমি একবার x-y নিয়ে, আরেকবার y-x নিয়ে দেখতে পারছি।

```python
fig = plt.Figure()
fig
```

```python
axes = fig.add_axes([.1, .1, .8, .8])

axes.plot(x,y)
fig
```

## custom figure ও axes কনফিগারেশন

এখানে `fig = plt.Figure()` দিয়ে একটি নতুন figure বানানো হয়েছে।

- `fig.add_axes([.1, .1, .8, .8])` এর মানে figure-এর ভিতরে plot-এর অবস্থান ও সাইজ দেওয়া হয়েছে।
- এখানে `axes` variable দিয়ে plot area নির্ধারণ করা হচ্ছে।
- `axes.plot(x, y)` মানে x ও y ডেটা দিয়ে একটি লাইন গ্রাফ তৈরি করা হচ্ছে।

এই অংশে আমরা শিখছি কিভাবে plot-এর size, position, এবং layout custom control করা যায়।

এটা `subplot`-এর থেকে একটু ভিন্ন, কারণ এখানে আমরা নিজের মতো figure-এ plotting area সাজাই।

```python
axes.plot(x,y)
fig
```

```python
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(15,5))
```

## multiple subplot তৈরি

এখানে `plt.subplots(nrows=1, ncols=3, figsize=(15,5))` ব্যবহার করা হয়েছে।

- ১ সারি ও ৩ কলাম নিয়ে ৩টি subplot তৈরি হয়েছে।
- `fig, axes` অর্থাৎ figure এবং subplot array দুইটাই একসাথে নেওয়া হয়েছে।
- `figsize=(15,5)` মানে figure-এর আকার দেওয়া হয়েছে।

এরপর `axes[2].plot(x,y)` ব্যবহার করে ৩য় subplot-এ plot করা হয়েছে।

এই অংশে আমরা শিখি কিভাবে একসাথে অনেক subplot তৈরি করে বিভিন্ন chart একসাথে দেখানো যায়।

```python
import numpy as np
import matplotlib.pyplot as plt


fig, axes = plt.subplots(nrows=1, ncols=3)
plt.tight_layout()
```

```python
axes
```

```python
axes[2].plot(x,y)
```

```python
fig
```

```python
fig = plt.Figure()
fig
```

```python
axes = fig.add_axes([0,0,1,1])
fig
```

```python
axes.plot(x,y, label = 'x,y')
axes.plot(x, x**2, label='x,x**2')
axes.legend()
```

## legend ও multiple plotting

এখানে `axes.plot(x, y, label='x,y')` এবং `axes.plot(x, x**2, label='x,x**2')` ব্যবহার করা হয়েছে।

- প্রতিটি line-এ `label` দেওয়া হয়েছে, যাতে legend-এ নাম দেখা যায়।
- `axes.legend()` দিয়ে plot-এ legend দেখানো হয়েছে।

এই অংশে আমরা শিখি কিভাবে একই axes-এ একাধিক curve plot করে তাদের নাম দেখানো যায়।

অর্থাৎ, এক plot-এ দুইটি আলাদা trend বা function একই সাথে তুলনা করা হচ্ছে।

```python
axes.plot(x,x**2)
fig
```
